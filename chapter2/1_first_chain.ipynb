# **Chains in LangChain**

```python
# it is a good thing to  make this file at the top of the notebook it's a good practise
from langchain_openai import ChatOpenAI
import os
from load_dotenv import load_dotenv

# this function load all the variables from the .env file and 
# will make them available in the os.environ dictionary (env variable)
load_dotenv()

if os.environ.get("OPENAI_API_KEY"):
    print("got the api key")
else:
    print("not got api key")
```

```python
# from openai.types import model
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

# it only give the content part of the output (task3)
from langchain_core.output_parsers import StrOutputParser

llm_ollama= ChatOllama(model='qwen2.5:3b', temperature=0)
my_messages = [
    SystemMessage(content="you are a gen-z assistant who always answer in fun way"),
    HumanMessage(content="hey")
]

llm_ollama.invoke(my_messages).content
```

## **First Chain**

```python
# prompt 1st TASK

prompt_template = ChatPromptTemplate([
    ('system','You are a helpful assistant'),
    ('human','{input}')
])
```

```python
# TASK 2 LLM

llm_ollama= ChatOllama(model='qwen2.5:3b', temperature=0)
```

```python
# TASK 3 Parsers

parcer = StrOutputParser()
```

### **Manual Invocation**

```python
template = prompt_template.invoke({'input':'is there really a chinease language or there just mandrid ?'})
res = llm_ollama.invoke(template)
final_res = res.content
final_res
```

### **Chain Invocation**

```python
chain = prompt_template | llm_ollama | parcer

chain.invoke({'input':"what is the capital of bulgaria ?"})
```

```python
from langchain_core.runnables import RunnableSequence

chain1 = RunnableSequence(prompt_template, llm_ollama, parcer)

chain1.invoke('name any city thats name is familiar to a persons name')
```
