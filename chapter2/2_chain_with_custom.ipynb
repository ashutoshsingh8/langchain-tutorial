# **Chain with Custom Runnable**

```python
# prompt 1st TASK

prompt_template = ChatPromptTemplate([
    ('system','You are a helpful assistant'),
    ('human','{input}')
])
```

```python
# TASK 2 LLM

llm_ollama= ChatOllama(model='qwen2.5:3b', temperature=0)
```

```python
# TASK 3 Parsers

parcer = StrOutputParser()
```

```python
# TASK 4 custom function

# In LangChain, a RunnableLambda acts as a bridge or a wrapper for your custom Python functions
"""
Why do we need it?

1... LangChain chains expect components to follow a specific interface called a Runnable. 
If you write a standard Python function to process data (like a custom dictionary creator), 
LangChain doesn't recognize it as part of a pipeline by default.

2... By wrapping your function in RunnableLambda, you explicitly tell LangChain to treat 
your custom logic as a valid "task" or "node" in your workflow.

Key points for your notes:

1... Function to Task: It converts a basic Python function into a component that can be 
used in an LCEL (LangChain Expression Language) chain.

2... Seamless Integration: Once wrapped, your function can be linked using the pipe (|) 
operator just like built-in components.

3... Modularity: It allows you to inject custom business logic (like reformatting data 
or calling external APIs) anywhere within your chain without breaking the flow.

"""

from langchain_core.runnables import RunnableLambda

def dict_maker(text: str)-> dict :
    return {"text": text}

dict_maker_runnable= RunnableLambda(dict_maker)
```

```python
# TASK 5 post prompt
 
prompt_post = ChatPromptTemplate.from_messages(
    messages=[
        ('system', 'you are a social media post generator'),
        ('human', 'create a post from following text for LinkedIn {text}'),
    ]
)
```

```python
# TASK 6 Generate the post

llm_ollama= ChatOllama(model='qwen2.5:3b', temperature=0)
```

```python
# TASK 7 string parser

parcer = StrOutputParser()
```

```python
custom_chain = prompt_template | llm_ollama | parcer | dict_maker_runnable | prompt_post | llm_ollama | parcer

custom_chain.invoke('what is the capital of france?')
```
