# **Using Pydantic Models**

```python
from pydantic import BaseModel, Field

# schema class that llm should follow while producing output

class llm_schema(BaseModel):
    setup: str = Field(description='The setup of the joke')
    punchline: str = Field(description='The punchline for the joke')
```

```python
obj = llm_schema(**{'setup':'some setup', 'punchline': 'some [unchlinmw]'})
obj
```

```python
# if we use llm_structure_output llm model it always produce the output in the given scructure
llm_structured_output = llm_ollama.with_structured_output(llm_schema)

result = llm_structured_output.invoke("Tell me a fact about doctor")
# result.punchline
result
```

# **Using TypedDict**

```python
from typing import TypedDict

class llm_schema_td(TypedDict):
    setup: str
    punchline: str
```

```python
llm_structured_output_td = llm_ollama.with_structured_output(llm_schema_td)
result_td = llm_structured_output_td.invoke('Tell me a random fact that is not true')
result_td

# give the output in dictionart not in pydantic object
```
