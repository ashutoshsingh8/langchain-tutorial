```python
import requests
from langchain_ollama import ChatOllama

# Ollama typically runs on localhost:11434
ollama_url = "http://localhost:11434"

try:
    # Send a request to check if the local Ollama server is active
    response = requests.get(ollama_url)
    if response.status_code == 200:
        print("Ollama server is running locally!")
        
        # Initialize the model (replace 'llama3' with your downloaded model)
        llm = ChatOllama(model="qwen2.5:3b", base_url=ollama_url)
        print("ChatOllama initialized successfully.")
    else:
        print(f"Ollama server responded with status code: {response.status_code}")
except requests.exceptions.ConnectionError:
    print("Ollama server is not running. Please start the Ollama application.")
```

# **PROMPTS**

Prompts are the messages that are send to the LLM. Prompts are more user friendly than messages.


```python
from langchain_core.prompts import PromptTemplate

user_input = input('Enter a topic for fun fact: ')

dynamic_prompt = PromptTemplate.from_template("Write a fun face about {topic}")

ready_prompts = dynamic_prompt.invoke({"topic": user_input})

# dynamic_prompt --> 
# PromptTemplate(input_variables=['topic'], input_types={}, partial_variables={}, template='Write a fun face about {topic}')
# dynamic_prompt.invoke({"topic": user_input}) --> 
# StringPromptValue(text='Write a fun face about independence')

llm_ollama.invoke(ready_prompts).content
```

```python
from langchain_core.prompts import ChatPromptTemplate
# we don't need to write system prompt, ai prompt, user prompt we only make a list of tuple having system and user and it automatically detects the system and the user prompts

prompt_template = ChatPromptTemplate.from_messages([
    ('system', 'You are a pol{tone} ite assistant'),
    ('user', 'Write a fun face about {topic}')
])

user_input = input("Enter the topic of the fun fact: ")
user_tone = input("Enter the tone for the llm: ")

ready_prompts = prompt_template.invoke({'topic':user_input, 'tone': user_tone})

llm_ollama.invoke(ready_prompts.messages).content

# ready_prompts --> 
# ChatPromptValue(messages=[SystemMessage(content='You are a polite assistant', additional_kwargs={}, response_metadata={}), HumanMessage(content='Write a fun face about AI', additional_kwargs={}, response_metadata={})])
# ready_prompts.messages -->
# [SystemMessage(content='You are a polite assistant', additional_kwargs={}, response_metadata={}),
#  HumanMessage(content='Write a fun face about AI', additional_kwargs={}, response_metadata={})]
```
